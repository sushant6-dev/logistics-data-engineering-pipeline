## Data Schema

The project uses the following shipment-level data schema:

| Column | Description |
|---|---|
| `shipment_id` | Unique identifier assigned to each shipment |
| `customer_id` | Unique identifier of the customer associated with the shipment |
| `origin_city` | City from which the shipment originates |
| `destination_city` | City where the shipment is destined |
| `origin_hub` | Origin logistics hub handling the shipment |
| `destination_hub` | Destination logistics hub handling the shipment |
| `flight_date` | Date on which the shipment is scheduled to depart |
| `eta_date` | Estimated date of arrival at the destination |
| `weight_kg` | Weight of the shipment in kilograms |
| `shipment_type` | Type of shipment, such as Apex, Express or Standard |
| `payment_mode` | Payment method used for the shipment, such as COD or Prepaid |
| `delivery_attempts` | Number of delivery attempts made for the shipment |
| `shipment_status` | Current status of the shipment, such as In Transit, Delivered, or Failed |
