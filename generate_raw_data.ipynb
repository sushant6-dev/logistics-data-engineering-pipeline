## Data Schema

The project uses the following shipment-level data schema:

| Column | Description |
|---|---|
| `shipment_id` | Unique identifier assigned to each shipment |
| `customer_id` | Unique identifier of the customer associated with the shipment |
| `origin_city` | City from which the shipment originates |
| `destination_city` | City where the shipment is destined |
| `origin_hub` | Origin logistics hub handling the shipment |
| `destination_hub` | Destination logistics hub handling the shipment |
| `flight_date` | Date on which the shipment is scheduled to depart |
| `eta_date` | Estimated date of arrival at the destination |
| `weight_kg` | Weight of the shipment in kilograms |
| `shipment_type` | Type of shipment, such as Apex, Express or Standard |
| `payment_mode` | Payment method used for the shipment, such as COD or Prepaid |
| `delivery_attempts` | Number of delivery attempts made for the shipment |
| `shipment_status` | Current status of the shipment, such as In Transit, Delivered, or Failed |

```python
DSTN_CITY = ['Pune', 'Mumbai', 'Banglore', 'Hyderabad', 'Thane', 'Navi Mumbai', 'Nagpur', 'Kolkata', 'UP', 'Assam']
ORGN_CITY = ['Pune', 'Mumbai', 'Banglore', 'Hyderabad', 'Thane', 'Navi Mumbai', 'Nagpur', 'Kolkata', 'UP', 'Odissa']

ORGN_HUB = ['PUN01', 'MUM01', 'BLR01', 'HYD01', 'THN01', 'NVM01', 'NGP01', 'KOL01', 'LKO01', 'GUW01']
DSTN_HUB = ['PUN01', 'MUM01', 'BLR01', 'HYD01', 'THN01', 'NVM01', 'NGP01', 'KOL01', 'LKO01', 'GUW01']

HUB_CODES = {
    'PUN01': 'Pune',
    'MUM01': 'Mumbai',
    'BLR01': 'Bangalore',
    'HYD01': 'Hyderabad',
    'THN01': 'Thane',
    'NVM01': 'Navi Mumbai',
    'NGP01': 'Nagpur',
    'KOL01': 'Kolkata',
    'LKO01': 'Lucknow',
    'GUW01': 'Guwahati'
}

shipment_type_list = ['Apex', 'Standard', 'Express']
payment_modes = ['Card', 'UPI', 'COD']
single_day = ['PUN01', 'NGP01', 'NVM01', 'THN01', 'MUM01']
raw_random_data = [{
    'shipment_id' : 'SHP00001',
    'customer_id' : 'C00001',
    'origin_city' : 'Mumbai',
    'destination_city' : 'Banglore',
    'origin_hub' : 'MUM01',
    'destination_hub' : 'BLR01',
    'flight_date' : '01-09-2026',
    'eta_date' : '03-09-2026',
    'weight_kg' : 3,
    'shipment_type' : 'Apex',
    'payment_mode' : 'Card',
    'shipment_status' : 'in-transit'
}]

print(raw_random_data[0])
```

```python
import random
import datetime as dt
print(random.choice(ORGN_HUB))
```

```python
raw_data = []

for i in range(1, 1001):
    
    # Generating Shipment IDs.
    shipment_id = f"SHP{i:04d}"
    
    # Selecting random origin and destination.
    orgn_hub = random.choice(ORGN_HUB)
    dstn_hub = random.choice(DSTN_HUB)
    
    # Generating Randome Dates for flight and delivery date.
    start_date = dt.date(day=20, month=8, year=2026)
    end_date = dt.date(day=20, month=9, year=2026)
    days_between = (end_date - start_date).days
    flt_date = start_date + dt.timedelta(random.randrange(days_between))
    eta_date = flt_date
    
    weight = random.randint(2,5)
     
    if orgn_hub in single_day and dstn_hub in single_day :
        eta_date = flt_date + dt.timedelta(1)
    else: 
        eta_date = flt_date + dt.timedelta(random.randrange(2,6))
    
    # Generating Random Shipment types
    shipment_type = random.choice(shipment_type_list)
    
    # Generating Shipment Status
    shipment_status = ''
    if eta_date > dt.date(day = 25, month=9, year=2026) : 
        shipment_status = 'In-Transit'
    else:
        shipment_status = random.choice(['Delivered', 'Failed'])
    
    
    # Generating Random Payment Modes.
    payment_md = random.choice(payment_modes)
        
    row_data = {
    'shipment_id' : shipment_id,
    'customer_id' : 'C00001',
    'origin_city' : HUB_CODES[orgn_hub],
    'destination_city' : HUB_CODES[dstn_hub],
    'origin_hub' : orgn_hub,
    'destination_hub' : dstn_hub,
    'flight_date' : flt_date,
    'eta_date' : eta_date,
    'weight_kg' : weight,
    'shipment_type' : shipment_type,
    'payment_mode' : payment_md,
    'shipment_status' : shipment_status
    }
    raw_random_data.append(row_data)
    
    
    
```
